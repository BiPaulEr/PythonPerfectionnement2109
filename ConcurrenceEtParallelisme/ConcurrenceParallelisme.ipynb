```python
from threading import Thread
import time

def worker_1(caractere):
    for i in range(0, 10):
        time.sleep(1)
        print(caractere, flush=True, end='')

t1 = Thread(target=worker_1, args="$")
t2 = Thread(target=worker_1, args="*")
t1.start()
t1.join()
t2.start()
print("MT3")
```
