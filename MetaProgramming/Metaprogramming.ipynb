```python
class Vector:
     def __init__(self, x, y):
         self.x = x
         self.y = y
     def __repr__(self):
         return f"<Vector: x={self.x}, y={self.y}>"
     def __eq__(self, other):
         return self.x == other.x and self.y == other.y
```

```python
v1 = Vector(3, 4)
v2 = Vector(6, 7)
v3 = Vector(3,4)
```

```python
print(v1)
```

```python
v1 == v2
```

```python
v1 == v3
```

```python
v1 == v1
```

```python
class Vector:
     def __init__(self, x, y, z):
         self.x = x
         self.y = y
         self.z = z
     def __repr__(self):
         return f"<Vector: x={self.x}, y={self.y}, z={self.z}>"
     def __eq__(self, other):
         return self.x == other.x and self.y == other.y and self.z == other.z
```

```python
v1 = Vector(3, 5, 6)
```

```python
from dataclasses import dataclass

@dataclass(frozen = True)
class VectorDataclass:
    x : int
    y : int
v1 = VectorDataclass(1, 2)
v1.x = 90
```

```python
from dataclasses import field

@dataclass
class ListeDEcourse:
    legumes : list = field(default_factory=list)
    fruits : list  =  field(default_factory=list)
v1 = ListeDEcourse([], [] )
v2 = ListeDEcourse()
print(v2)
```

```python
from dataclasses import field

@dataclass(order=True)
class ProduitCourse:
    prix : float = field(compare=False)
    name : str
p1 = ProduitCourse(5.5, "Orange" )
p2 = ProduitCourse(8.9, "Apple")

print(p1 > p2)
print(p2 > p1)
```

```python
from dataclasses import field

@dataclass
class ListeDeCourse:
    legumes : list 
    fruits : list
    total_prix : float = field(init=False)
    def __post_init__(self):
        self.total_prix = sum(self.legumes + self.fruits)
ldc1 = ListeDeCourse([4], [7, 8])
```
