```python
class ContextManager:
    def __enter__(self):
        print("enter context")
        return "OBEJCT DISPONIBLE DANS LE CONTEXTE"
        
    def __exit__(self, type_exception, instance_exception, traceback):
        print(type_exception)
        print(instance_exception)
        if type_exception is None:
            print("PAS DERREUR")
        if type_exception is not None:
            print("ERREUR DETECTE")
        if type_exception is StopIteration:
            print("StopIteration DETECTE")
        print("leave the contexte")
        return False

with ContextManager() as contexte:
    print(contexte)
    raise SyntaxError("ERUURE")
    print("JE suis dans le contexte")
```
