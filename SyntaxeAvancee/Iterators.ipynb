```python
class CountDown:
    def __init__(self, init):
        self.step = init
        
    def __next__(self):
        if self.step <= 0:
            raise StopIteration
        self.step = self.step -1
        return self.step²

    def __iter__(self):
        return self

countDown = CountDown(11)
for i in countDown:
    print(i)
```

```python
class CountDownState:
    def __init__(self, init):
        self.step = init
        
    def __next__(self):
        if self.step <= 0:
            raise StopIteration
        self.step = self.step -1
        return self.step
        
class ClassDown:
    def __init__(self, init):
            self.step = init
    def __iter__(self):
        return CountDownState(self.step)

countDown = ClassDown(11)
for i in countDown:
    print(i)
for i in countDown:
    print(i)
```

```python
class Point:
    def __init__(self, x, y):
       self.x = x
       self.y = y
    def __repr__(self):
        return f"Point<{self.x},{self.y}>"
       
p = Point(3, 5)
print(p)
```
