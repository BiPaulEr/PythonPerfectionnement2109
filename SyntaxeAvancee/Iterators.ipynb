```python
class CountDown:
    def __init__(self, init):
        self.step = init
        
    def __next__(self):
        if self.step <= 0:
            raise StopIteration
        self.step = self.step -1
        return self.step²

    def __iter__(self):
        return self

countDown = CountDown(11)
for i in countDown:
    print(i)
```

```python
class CountDownState:
    def __init__(self, init):
        self.step = init
        
    def __next__(self):
        if self.step <= 0:
            raise StopIteration
        self.step = self.step -1
        return self.step
        
class ClassDown:
    def __init__(self, init):
            self.step = init
    def __iter__(self):
        return CountDownState(self.step)

countDown = ClassDown(11)
for i in countDown:
    print(i)
for i in countDown:
    print(i)
```

```python
class PrimeState:
    def __init__(self, limit):
        self.limit = limit
        self.step = 2
        
    def __next__(self):
        while self.step <= self.limit:
            is_prime = True
            for i in range(2, self.step):
                if self.step % i == 0:
                    is_prime = False
                    break
            if is_prime:
                self.step += 1
                return self.step - 1
            self.step += 1
        raise StopIteration
        
                
        
class PrimeIterator:
    def __init__(self, init):
            self.step = init
    def __iter__(self):
        return PrimeState(self.step)

prime = PrimeIterator(29)
for i in prime:
    print(i)
```

```python
def ClassDownGenerator(init):       
    for i in range(init, -1, -1):
        yield i
        

countDown = ClassDownGenerator(11)
for i in countDown:
    print(i)
for i in countDown:
    print(i)
```
