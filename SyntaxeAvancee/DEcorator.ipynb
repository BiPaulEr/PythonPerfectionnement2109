```python
import time
def benchamrk(function):
    def wrapper():
        begin = time.time()
        function()
        end = time.time() - begin
        print(f"{end} ms")
    return wrapper

def function1():
    time.sleep(1)
    print("function")

def function2():
    time.sleep(3)
    print("function2")

function1_bench = benchamrk(function1)
function2_bench = benchamrk(function2)
```

```python
import time
def benchamrk1(function):
    def wrapper():
        begin = time.time()
        function()
        end = time.time() - begin
        print(f"1-> {end} ms")
    return wrapper
    
def benchamrk2(function):
    def wrapper():
        begin = time.time()
        function()
        end = time.time() - begin
        print(f"2-> {end} ms")
    return wrapper
    
@benchamrk1
def function_test1():
    time.sleep(1)
    print("function")
    

@benchamrk2
def function_test2():
    time.sleep(3)
    print("function2")

function_test2()
function_test1()
```

```python
import time
def cache(function):
    result = {}
    def wrapper(*args):
        if args in result:
            print("cached !")
            return result[args]
        resultat = function(*args)
        result[args] = resultat
        print(result)
        return resultat
    return wrapper

@cache
def function_lente(a, b):
    time.sleep(2)
    return a + b
    
@cache
def function_lente2(a, b):
    time.sleep(1)
    return a + b
```

```python
print(function_lente(4, 5))
print(function_lente(7, 8))
print(function_lente(4, 5))
print(function_lente(7, 8))
```

```python
def time_execution(function):
    def wrapper(*args, **kwargs):
        begin = time.time()
        result = function(*args, **kwargs)
        end = time.time()
        print(f" {function.__name__}: {end-begin:.2f}s")
        return result
    return wrapper


@time_execution
def fetch_data():
    time.sleep(2)  # Simule une opération prenant du temps
    return "Data"

fetch_data()
```

```python
def time_execution_personalise(nom_personnalise):
    def time_execution(function):
        def wrapper(*args, **kwargs):
            begin = time.time()
            result = function(*args, **kwargs)
            end = time.time()
            print(f"{nom_personnalise}-> {function.__name__}: {end-begin:.2f}s")
            return result
        return wrapper
    return time_execution


@time_execution_personalise("CA MARCHE ?")
def fetch_data():
    time.sleep(2)  # Simule une opération prenant du temps
    return "Data"

fetch_data()
```
